```python
import os
import logging
import sys

from ska_integration_test_harness.config.reader import YAMLConfigurationReader
from ska_integration_test_harness.emulated import EmulatedSDPWrapper
from ska_integration_test_harness.structure import TelescopeWrapper
from ska_low_csp_testware import LowCSPWrapper, LowCSPFacade, LowTMCWrapper, NoDishesWrapper
from ska_integration_test_harness.inputs import FileJSONInput
```

```python
os.environ["TANGO_HOST"] = "ska-low-csp-databaseds.ska-low-csp-dev.svc.cluster.local:10000"
```

```python
logging.basicConfig(
    format="%(asctime)s | %(levelname)s : %(message)s",
    level=logging.INFO,
    stream=sys.stdout,
)
```

```python
config_reader = YAMLConfigurationReader()
config_reader.read_configuration_file("../resources/test_harness_config.yml")
config = config_reader.get_test_harness_configuration()
```

```python
telescope = TelescopeWrapper()
telescope.set_up(
    tmc=LowTMCWrapper(config.tmc_config),
    csp=LowCSPWrapper(config.csp_config),
    sdp=EmulatedSDPWrapper(config.sdp_config),
    dishes=NoDishesWrapper(config.dishes_config),
)
```

```python
low_csp_facade = LowCSPFacade(telescope)
```

```python
low_csp_facade.admin_mode_online()
low_csp_facade.assign_resources(input=FileJSONInput("../resources/assign_resources.json"))
low_csp_facade.configure(input=FileJSONInput("../resources/configure.json"))
low_csp_facade.abort()
low_csp_facade.restart()
low_csp_facade.admin_mode_offline()
```
